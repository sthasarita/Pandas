```python
# We generally do NOT replace missing (NaN) values with 0,
# because 0 may not represent the actual data and can distort the analysis.

# Instead, we fill missing vnba alues using:
# 1. Mean
# 2. Median
# 3. Mode
# The choice depends on the type and distribution of the data.

# ---------------------- Mean ----------------------
# Use the mean when:
# - The data is numerical.
# - The distribution is approximately normal (symmetric).
# - There are no significant outliers.

# Example:
# 20000, 30000, 40000
# Mean = (20000 + 30000 + 40000) / 3 = 30000
# Since there are no extreme values, the mean is a good representative.

# ---------------------- Median ----------------------
# Use the median when:
# - The data is numerical.
# - The distribution is skewed (left or right skew).
# - There are outliers.

# Example:
# 20000, 60000, 300000
# Mean   = 126666.67 (heavily affected by the outlier)
# Median = 60000 (better represents the typical value)

# Therefore, for skewed data with outliers, the median is preferred.

# ---------------------- Mode ----------------------
# Use the mode when:
# - The data is categorical (text labels).
# - Or when you want to fill missing values with the most frequently occurring value.

# Example:
# Color:
# Red, Blue, Red, Green, Red, NaN
# Mode = Red
# Replace NaN with "Red".

# Summary:
# - Mean   -> Numerical data, normal distribution, no outliers.
# - Median -> Numerical data, skewed distribution or presence of outliers.
# - Mode   -> Categorical data (or the most frequent value in any column).
```

## Sort a DataFrame with the sort_values Method II
- To sort by multiple columns, pass the `by` parameter a list of column names. Pandas will sort in the specified column order (first to last).
- Pass the `ascending` parameter a Boolean to sort all columns in a consistent order (all ascending or all descending).
- Pass `ascending` a list to customize the sort order *per* column. The `ascending` list length must match the `by` list.

```python
nba = pd.read_csv("nba.csv")
nba.tail()
```

```python
nba.sort_values(["Team", "Name"])
nba.sort_values(by=["Team", "Name"])
nba.sort_values(by=["Team", "Name"], ascending=True)
nba.sort_values(by=["Team", "Name"], ascending=False)
```

```python
nba.sort_values(by=["Team", "Name"], ascending=[True, False])

nba.sort_values(["Position", "Salary"])
nba.sort_values(["Position", "Salary"], ascending=True)
nba.sort_values(["Position", "Salary"], ascending=False)
nba.sort_values(["Position", "Salary"], ascending=[True, False])
nba.sort_values(["Position", "Salary"], ascending=[False, True])

nba = nba.sort_values(["Position", "Salary"], ascending=[False, True])
nba
```

## Rank Values with the rank Method

```python
nba["Salary"] = nba["Salary"].sort_values(ascending = True)

nba["Salary"]
# because pandas matched values by their index labels during assignment,
# not by their position. The values belonging to index 0, 1, 2, etc., were
# placed back into those same rows, so the first few rows appeared 
# unchanged. This behavior—automatic alignment by index—is one of pandas'
# most important features to understand.
```

```python
# the output are the ranks

nba["Salary"].rank(ascending = True).head()
```
