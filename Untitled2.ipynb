```python
import numpy as np

arr = np.arange(12)
print("Original:", arr)

#Slicing
print("First 5:", arr[:5])
print("Last 3:", arr[-3:])

# reshaping we can reshape the numpy array according to our wish but in such a way that the total number in each reshaped matrix must be same
reshaped = arr.reshape(3,4)
print("Reshaped 3x4 :\n",reshaped)

# reshaped[a,b] here a means rows and b means columns so we can access the required rows and columns according to our choice
# reshaped[a] means simple the rows we are indexing 
print("\nLast Column = \n", reshaped[:,-1])
print("\nLast Row = \n", reshaped[-1])

# numpy array is a data structure containing only numbers as its elements 

reshaped[0][-1]  = 55

print("After changing the data of [0][-1] position, the reshaped array is ", reshaped)
```
