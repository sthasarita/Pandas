```python
ch["Salary"] = ch["Salary"].str.replace("$", "", regex = False).str.replace(",", "", regex = False).astype("float")
```

```python
ch.head()
```

```python
columns_to_select = ["Name", "Position Title", "Department"]
```

```python
for column in columns_to_select:
    ch[column] = ch[column].str.title()
```

```python
ch.groupby("Department")["Salary"].agg([
    "min", 
    "max", 
    "mean",
    "median",
    ("Count", "count"),
    ("Range" , lambda x : x.max() - x.min())
])
```

```python
ch.groupby("Department")["Salary"].agg([
    ("Sum", "sum"),
    ("Average", lambda x : x.sum()/x.count()),
    ("Minimum_Value", "min"), 
    ("Maximum_Value", "max"),
    ("Median", "median"),
    ("Range" , lambda x: x.max() - x.min())
])
```

```python
ch["dept_median_salary"] = ch.groupby("Department")["Salary"].transform("median")
```

```python
def top_employees(group, n = 5):
    return group.nlargest(n, "Salary")


top_by_pos = ch.groupby("Position Title").apply(top_employees, n = 3)
```

```python
top_by_pos.reset_index(inplace = True)
```
