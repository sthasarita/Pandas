# Merging dataframes

```python
import numpy as np
# Create sample dataframes for merging
students = pd.DataFrame({
    'StudentID': [101, 102, 103, 104, 105, 106],
    'Name': ['John', 'Jane', 'Bob', 'Alice', 'Charlie', 'Henry'],
    'SubjectID': [1, 2, 1, 3, 2, np.nan]
})

subjects = pd.DataFrame({
    'SubjectID': [1, 2, 3, 4],
    'SubName': ['AI', 'OS', 'TOC', 'DBMS'],
    
})

marks = pd.DataFrame({
    'StudentID': [101, 102, 103, 104, 105,106],
    'marks': [60, 75, 55, 80, 70,80],
})

print("Students:")
print(students)
print("\nSubjects:")
print(subjects)
print("\nMarks:")
print(marks)
```

```python
inner_join = pd.merge(students, subjects, on='SubjectID', how='inner')
inner_join
```

```python
left_join = pd.merge(students, subjects, on='SubjectID', how='left')
left_join
```

```python
right_join = pd.merge(students, subjects, on='SubjectID', how='right')
right_join
```

```python

outer_join = pd.merge(students, subjects, on='SubjectID', how='outer')
outer_join
```

```python
merged = students.merge(subjects, on='SubjectID').merge(marks, on='StudentID')
merged
```

# concatenation

```python
df1 = pd.DataFrame({
    "ID": [1, 2],
    "Name": ["Anita", "Alice"]
})

df2 = pd.DataFrame({
    "ID": [3, 4],
    "Name": ["Bibek", "Rahul"]
})
```

```python
result = pd.concat([df1, df2])
```

```python
result
```

```python
result = pd.concat([df1, df2], ignore_index = True)
```
