# Series

## Create a Series Object from a List
- A pandas **Series** is a one-dimensional labelled array.
- A Series combines the best features of a list and a dictionary.
- A Series maintains a single collection of ordered values (i.e. a single column of data).
- We can assign each value an identifier, which does not have to *be* unique.

```python
# creating the series from the list
ice_cream = ["Chocolate", "Vanilla", "Strawberry", "Rum Raisin"]
pd.Series(ice_cream)
```

```python
# pd.series() has second attribute index where we can send the index for the series
ice_cream = ["Chocolate", "Vanilla", "Strawberry", "Rum Raisin"]
ice_cream = pd.Series(ice_cream, index = ["Chocolate", "Vanilla", "Strawberry", "Rum Raisin"])
ice_cream
```

```python
# accessing the series value through indexing
print(ice_cream["Chocolate"])
```

```python
# creating the series containing integers 
lottery_numbers = [4, 8, 15, 16, 23, 42]
pd.Series(lottery_numbers)
```

```python
# creating the series containing booleans
registrations = [True, False, False, False, True]
pd.Series(registrations)
```

## The sort_values Method
- Pandas series can be sorted using python function list however doing so doesn't changes the origingal series
- the sort_values also do not changes the original series
- The `sort_values` method sorts a **Series** values in order.
- By default, pandas applies an ascending sort (smallest to largest).
- Customize the sort order with the `ascending` parameter.

```python
pokemon = pd.read_csv("pokemon.csv", usecols=["Name"]).squeeze("columns")
google = pd.read_csv("google_stock_price.csv", usecols=["1. open"]).squeeze("columns")

google.head()
```

```python
print(google.sort_values())
print(google.sort_values(ascending=True))
print(google.sort_values(ascending=False))
print(google.sort_values(ascending=False).head())
```

```python
print(pokemon.sort_values())
print(pokemon.sort_values(ascending=True))
print(pokemon.sort_values(ascending=False))
print(pokemon.sort_values(ascending=True).tail())
```

## Extract Series Value by Index Position
- Use the `iloc` accessor to extract a **Series** value by its index position.
- `iloc` is short for "index location".
- Python's list slicing syntaxes (slices, slices from start, slices to end, etc.) are supported with **Series** objects.

```python
pokemon = pd.read_csv("pokemon.csv", usecols=["Name"]).squeeze("columns")
pokemon.head()
```

```python
print(pokemon.iloc[0])
print(pokemon.iloc[500])

# pokemon.iloc[1500])

print(pokemon.iloc[[100, 200, 300]])

# pokemon.iloc[[100, 200, 300, 1500]])

print(pokemon.iloc[27:36])
print(pokemon.iloc[0:7])
print(pokemon.iloc[:7])

print(pokemon.iloc[700:1010])
print(pokemon.iloc[700:])
print(pokemon.iloc[700:5000])

print(pokemon.iloc[-1])
print(pokemon.iloc[-10])

print(pokemon.iloc[-10:-5])
print(pokemon.iloc[-8:])
```

## The get Method on a Series
- The `get` method extracts a **Series** value by index label. It is an alternative option to square brackets.
- The `get` method's second argument sets the fallback value to return if the label/position does not exist.

```python
pokemon = pd.read_csv("pokemon.csv", index_col="Name").squeeze("columns")
pokemon.head()
```

```python
print(pokemon.get("Moltres"))
print(pokemon.loc["Moltres"])

# pokemon.loc["Digimon"]
print(pokemon.get("Digimon"))
print(pokemon.get("Digimon", "Nonexistent"))#digimon doesn't exist so it sets nonexistent in its values and creates new index digimon as well
                                             # and returns the same set value
print(pokemon.get("Moltres", "Nonexistent"))

print(pokemon.get(["Moltres", "Digimon"], "One of the values in the list was not found so setting the fallback value which sets the value and returns the same value"))
```

## Math Methods on Series Objects
- The `count` method returns the number of values in the **Series**. It excludes missing values; the `size` attribute includes missing values.
- The `sum` method adds together the **Series's** values.
- The `product` method multiplies together the **Series's** values.
- The `mean` method calculates the average of the **Series's** values.
- The `std` method calculates the standard deviation of the **Series's** values.
- The `max` method returns the largest value in the **Series**.
- The `min` method returns the smallest value in the **Series**.
- The `median` method returns the median of the **Series** (the value in the middle).
- The `mode` method returns the mode of the **Series** (the most frequent alue).
- The `describe` method returns a summary with various mathematical calculations
- .The `info` method displays a summary of the Series, including its index, number of non-null values, data type, memory usage, and the Series name (if available).

```python
google = pd.read_csv("google_stock_price.csv", usecols=["1. open"]).squeeze("columns")
google.head()
```

```python
print(google.count())
print(google.sum())
print(google.product())
print(pd.Series([1, 2, 3, 4]).product())
print(google.mean())
print(google.std())
```

```python
print(google.max())
print(google.min())
print(google.median())
print(google.mode())
print(pd.Series([1, 2, 2, 2, 3]).mode())

print(google.describe())
print(google.info())
```

## Broadcasting
- **Broadcasting** describes the process of applying an arithmetic operation to an array (i.e., a **Series**).
- We can combine mathematical operators with a **Series** to apply the mathematical operation to every value.
- There are also methods to accomplish the same results (`add`, `sub`, `mul`, `div`, etc.)

```python
google = pd.read_csv("google_stock_price.csv", usecols=["1. open"]).squeeze("columns")
google.head()
```

```python
print(google.add(10))
print(google + 10)

print(google.sub(30))
print(google - 30)

print(google.mul(1.25))
print(google * 1.25)
print(1.25 * google)

print(google.div(2))
print(google / 2)
```

## The value_counts Method
- The `value_counts` method returns the number of times each unique value occurs in the **Series**.
- The `normalize` parameter returns the relative frequencies/percentages of the values instead of the counts.

```python
pokemon = pd.read_csv("pokemon.csv", index_col="Name").squeeze("columns")
pokemon.head()
```

```python
print(pokemon.value_counts())
print(pokemon.value_counts(ascending=True))
print(pokemon.value_counts(normalize=True))
print(pokemon.value_counts(normalize=True) * 100)
```

## The map Method
- The `map` method "maps" or connects each **Series** values to another value.
- We can pass the method a dictionary or a **Series**. Both types connects keys to values.
- The `map` method uses our argument to connect or bridge together the values.

```python
pokemon = pd.read_csv("pokemon.csv", index_col="Name").squeeze("columns")
pokemon
```

```python
# creating another series whose keys are same as the value of the pokemon series
attack_powers = pd.Series({
    "Grass, Poison": 10,
    "Fire": 15,
    "Water": 15,
    "Fairy, Fighting": 20,
    "Grass, Psychic": 50
})

attack_powers
```

```python
# map method sees whether the key value of attack_powers are same as the values of the series pokemon
# and if yes then it sets the new value i.e the values of the corresponding matched key in the place of that matched value
pokemon.map(attack_powers)
```
