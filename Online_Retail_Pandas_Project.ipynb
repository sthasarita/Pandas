```python
missing_cust = df[df["CustomerID"].isnull()]

# 1. Country breakdown (raw counts vs. percentage of that country's rows)
country_summary = pd.DataFrame({
    "Missing_Count": missing_cust["Country"].value_counts(),
    "Total_Count": df["Country"].value_counts(),
    "Missing_%": (missing_cust["Country"].value_counts() / df["Country"].value_counts() * 100).round(2)
}).dropna().sort_values("Missing_Count", ascending=False)

print("--- Missing CustomerID by Country ---")
print(country_summary)

# 2. Check InvoiceNo prefixes (e.g., 'C' for cancellations, 'A' for bad debt adjustments)
print("\n--- InvoiceNo Prefixes in Missing CustomerID ---")
print(missing_cust["InvoiceNo"].astype(str).str[0].value_counts())

# 3. Check for $0 UnitPrice and missing Descriptions inside this subset
print("\n--- Anomalies within Missing CustomerID rows ---")
print("Rows with UnitPrice == 0:", (missing_cust["UnitPrice"] == 0).sum())
print("Rows with UnitPrice < 0:", (missing_cust["UnitPrice"] < 0).sum())
print("Rows with missing Description:", missing_cust["Description"].isnull().sum())

# 4. Top descriptions or non-product StockCodes (e.g., DOT, POST, M, AMAZONFEE)
print("\n--- Top StockCodes in Missing CustomerID ---")
print(missing_cust["StockCode"].value_counts().head(10))
```

```python
# 1. Check unique StockCodes among the 1,454 missing Description rows
missing_desc_codes = df[df["Description"].isnull()]["StockCode"].unique()
print("Unique StockCodes with missing Description:", len(missing_desc_codes))

# 2. Build a lookup dictionary mapping each StockCode to its most frequent (mode) Description
valid_desc = df.dropna(subset=["Description"])
code_to_desc = (
    valid_desc.groupby("StockCode")["Description"]
    .agg(lambda x: x.mode().iloc[0])
)

# 3. See which StockCodes CANNOT be recovered (not present in valid_desc)
unrecoverable_codes = set(missing_desc_codes) - set(code_to_desc.index)
print("Unrecoverable StockCodes count:", len(unrecoverable_codes))
print("Rows affected by unrecoverable StockCodes:", df["StockCode"].isin(unrecoverable_codes).sum())

# 4. Recover missing descriptions using .map() and .fillna()
df["Description_recovered"] = df["Description"].fillna(df["StockCode"].map(code_to_desc))

print("Missing descriptions BEFORE recovery:", df["Description"].isnull().sum())
print("Missing descriptions AFTER recovery:", df["Description_recovered"].isnull().sum())
```
